```python
class Shape:
    def area(self):
        return 0
class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h
class Circle(Shape):
    PI = 3.14
    def __init__(self, r):
        self.r = r
    def area(self):
        return self.PI * self.r * self.r
def solve():
    shape_type = input().strip().upper()
    if shape_type == "RECTANGLE":
        w, h = map(float, input().split())
        shape = Rectangle(w, h)
    elif shape_type == "CIRCLE":
        r = float(input())
        shape = Circle(r)
    else:
        print("No")
        return

    print(f"{shape.area():.2f}")

solve()
```

```python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed
    def info(self):
        return f"{self.brand}-{self.speed}"
class Car(Vehicle):
    def __init__(self, brand, speed, seats):
        super().__init__(brand, speed)
        self.seats = seats

    def info(self):
        return f"{self.brand}-{self.speed}-{self.seats}"
def solve():
    obj_type = input().strip()
    data     = input().split()

    if obj_type == "Vehicle":
        brand, speed = data[0], data[1]
        obj = Vehicle(brand, speed)
    elif obj_type == "Car":
        brand, speed, seats = data[0], data[1], data[2]
        obj = Car(brand, speed, seats)
    else:
        print("No")
        return

    print(obj.info())

solve()
```

```python
class Polygon:
    def perimeter(self):
        return 0
class Triangle(Polygon):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    def perimeter(self):
        return self.a + self.b + self.c
def solve():
    a, b, c = map(int, input().split())
    triangle = Triangle(a, b, c)
    print(triangle.perimeter())

solve()
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age  = age

    def describe(self):
        return f"{self.name} {self.age}"

class Student(Person):
    def __init__(self, name, age, school):
        super().__init__(name, age)
        self.school = school

    def describe(self):
        return f"{self.name} {self.age} {self.school}"

def solve():
    obj_type = input().strip()
    data     = input().split()

    if obj_type == "Person":
        name, age = data[0], data[1]
        obj = Person(name, age)
    elif obj_type == "Student":
        name, age, school = data[0], data[1], data[2]
        obj = Student(name, age, school)
    else:
        print("No")
        return

    print(obj.describe())

solve()
```

#W8A20

```python
class Account:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, x):
        self.balance += x

    def get_balance(self):
        return self.balance

class SavingAccount(Account):
    def __init__(self, balance, rate):
        super().__init__(balance)
        self.rate = rate

    def add_interest(self):
        self.deposit(self.balance * self.rate)

def solve():
    balance, rate = map(float, input().split())
    acc = SavingAccount(balance, rate)
    acc.add_interest()
    print(f"{acc.get_balance():.2f}")

solve()
```

```python
class Matrix2x2:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d

    def trace(self):
        return self.a + self.d

    def determinant(self):
        return self.a * self.d - self.b * self.c

def solve():
    a, b, c, d = map(int, input().split())
    m = Matrix2x2(a, b, c, d)
    print(m.trace(), m.determinant())

solve()
```

#W8A25

```python
class PolynomialSum:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d

    def sum_str(self):
        k = self.a + self.c   
        m = self.b + self.d   

        if k == 0 and m == 0:
            return "0"
        if k == 0:
            return f"{m}"
        if m == 0:
            return f"{k}x"

        if m > 0:
            return f"{k}x + {m}"
        else:
            return f"{k}x - {abs(m)}"

def solve():
    a, b, c, d = map(int, input().split())
    p = PolynomialSum(a, b, c, d)
    print(p.sum_str())

solve()
```

```python
class Product:
    def __init__(self, name, price):
        self.name  = name
        self.price = price

    def final_price(self):
        return self.price

class DiscountProduct(Product):
    def __init__(self, name, price, percent):
        super().__init__(name, price)
        self.percent = percent

    def final_price(self):
        return self.price * (100 - self.percent) / 100

def solve():
    data    = input().split()
    name    = data[0]
    price   = float(data[1])
    percent = float(data[2])

    p = DiscountProduct(name, price, percent)
    print(f"{p.final_price():.2f}")

solve()
```

```python
class Counter:
    def __init__(self):
        self.count = 0

    def inc(self):
        self.count += 1

    def dec(self):
        self.count -= 1

    def value(self):
        return self.count

def solve():
    n  = int(input())
    ops = input().split()

    c = Counter()
    for op in ops:
        if op == "INC":
            c.inc()
        elif op == "DEC":
            c.dec()

    print(c.value())

solve()
```

```python
class Time:
    def __init__(self, h, m, s):
        self.h = h
        self.m = m
        self.s = s

    def to_seconds(self):
        return self.h * 3600 + self.m * 60 + self.s

def solve():
    h, m, s = map(int, input().split())
    t = Time(h, m, s)
    print(t.to_seconds())

solve()
```

```python
class Stack:
    def __init__(self):
        self.data = []

    def push(self, x):
        self.data.append(x)

    def pop(self):
        self.data.pop()

    def top(self):
        return self.data[-1]

def solve():
    n = int(input())
    s = Stack()

    for _ in range(n):
        line = input().split()
        if line[0] == "PUSH":
            s.push(int(line[1]))
        elif line[0] == "POP":
            s.pop()
        elif line[0] == "TOP":
            print(s.top())

solve()
```
