```python
from collections import Counter

def is_beautiful(arr):
    count = Counter(arr)
    return all(x == cnt for x, cnt in count.items())

def make_beautiful(arr):
    count = Counter(arr)
    removals = 0
    result = []

    for x, cnt in count.items():
        if x == 0:
            removals += cnt
        elif cnt == x:
            result.extend([x] * x)
        elif cnt > x:
            removals += cnt - x
            result.extend([x] * x)
        else:
            removals += cnt

    return removals, sorted(result)

def solve():
    n = int(input())
    arr = list(map(int, input().split()))

    if is_beautiful(arr):
        print(arr)
    else:
        removals, result = make_beautiful(arr)
        print(removals)
        print(result)

solve()
```

```python
def find_majority(arr):
    n = len(arr)
    if n == 0:
        return

    candidate = arr[0]
    count = 1

    for i in range(1, n):
        if count == 0:
            candidate = arr[i]
            count = 1
        elif arr[i] == candidate:
            count += 1
        else:
            count -= 1
    
    verify = 0
    for x in arr:
        if x == candidate:
            verify += 1

    if verify > n // 2:
        print(candidate)

def solve():
    n = int(input())
    arr = list(map(int, input().split()))
    find_majority(arr)

solve()
```

```python
def conditional_sort(arr):
    even = sorted([x for x in arr if x % 2 == 0])      
    odd  = sorted([x for x in arr if x % 2 != 0], reverse=True)  
    return even + odd

def solve():
    n = int(input())
    arr = list(map(int, input().split()))
    result = conditional_sort(arr)
    print(*result)

solve()
```

```python
def detect_order(arr):
    if len(arr) <= 1:
        return 'asc'
    for i in range(len(arr) - 1):
        if arr[i] < arr[i+1]:
            return 'asc'
        elif arr[i] > arr[i+1]:
            return 'desc'
    return 'asc' 

def merge_ascending(a, b):
    result = []
    i, j = 0, 0
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i]); i += 1
        else:
            result.append(b[j]); j += 1
    while i < len(a):
        result.append(a[i]); i += 1
    while j < len(b):
        result.append(b[j]); j += 1
    return result

def merge_sorted_arrays(a, b):
    order_a = detect_order(a)
    order_b = detect_order(b)

    a_asc = a[::-1] if order_a == 'desc' else a
    b_asc = b[::-1] if order_b == 'desc' else b

    merged = merge_ascending(a_asc, b_asc)

    if order_a == 'desc' and order_b == 'desc':
        return merged[::-1]

    return merged 

def solve():
    n = int(input())
    a = list(map(int, input().split()))
    m = int(input())
    b = list(map(int, input().split()))

    result = merge_sorted_arrays(a, b)
    print(*result)

solve()
```

```python
def is_leap_year(year):
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

def day_of_year(n, t, year=2020):
    days_in_month = [0, 31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    
    if is_leap_year(year):
        days_in_month[2] = 29

    if t < 1 or t > 12:
        print("No")
        return
    if n < 1 or n > days_in_month[t]:
        print("No")
        return

    total = sum(days_in_month[1:t]) + n
    print(total)

def solve():
    n, t = map(int, input().split())
    day_of_year(n, t, 2020)

solve()
```

```python
from collections import Counter

def solve():
    n = int(input())
    names = []
    for _ in range(n):
        full_name = input().strip()
        first_name = full_name.split()[-1]
        names.append(first_name)

    count = Counter(names)
    max_freq = max(count.values())

    result = min(name for name, freq in count.items() if freq == max_freq)
    print(result)

solve()
```

```python

def is_name(word):
    if len(word) < 2:
        return False
    if not word.isalpha():
        return False
    if not word[0].isupper():
        return False
    if not word[1:].islower():
        return False
    return True

def solve():
    s = input()
    words = s.split()
    count = sum(1 for w in words if is_name(w))
    print(count)

solve()
```

#W7A24

```python
def solve():
    s = input().strip()
    k = int(input())
    n = len(s)

    if k <= 0 or k > n:
        print("Không hợp lệ")
        return

    remove = n - k  
    stack = []

    for digit in s:
        while remove > 0 and stack and stack[-1] < digit:
            stack.pop()
            remove -= 1
        stack.append(digit)

    result = stack[:k]
    print(''.join(result))

solve()
```

#W7A25

```python
import re

def solve():
    text = input()
    
    numbers = re.findall(r'-?\d+\.?\d*', text)
    numbers = [float(n) for n in numbers]
    
    if not numbers:
        print("No")
        return
    
    total = sum(numbers)
    average = total / len(numbers)
    
    def fmt(x):
        return int(x) if x == int(x) else x
    
    print(f"{[fmt(n) for n in numbers]}")
    print(f"{fmt(total)}")
    print(f"{fmt(average)}")

solve()
```

```python
def solve():
    s = input().strip()
    stack = []
    pairs = {')': '(', '}': '{', ']': '['}

    for ch in s:
        if ch in '({[':
            stack.append(ch)
        elif ch in ')}]':
            if not stack or stack[-1] != pairs[ch]:
                print("No")
                return
            stack.pop()

    print("Yes" if not stack else "No")

solve()
```

```python
def solve():
    N = int(input())
    LIMIT = 10**6

    is_prime = [True] * (LIMIT + 1)
    is_prime[0] = is_prime[1] = False
    for i in range(2, int(LIMIT**0.5) + 1):
        if is_prime[i]:
            for j in range(i*i, LIMIT + 1, i):
                is_prime[j] = False

    primes = [p for p in range(max(2, N + 1), LIMIT + 1) if is_prime[p]]

    if len(primes) < 2:
        print("No")
        return

    best_len   = 0
    best_start = -1

    for i in range(len(primes) - 1):
        p, q = primes[i], primes[i + 1]
        gap = q - p - 1          
        if gap > best_len:       
            best_len   = gap
            best_start = p + 1

    if best_len == 0:
        print("No")
    else:
        result = list(range(best_start, best_start + best_len))
        print(f"{best_len}")
        print(f"{result}")

solve()
```

```python
def solve():
    arr = list(map(int, input().split()))
    n = len(arr)

    if n == 0:
        print("No")
        return

    best_start, best_len = 0, 1
    cur_start,  cur_len  = 0, 1

    for i in range(1, n):
        if arr[i] > arr[i - 1]:
            cur_len += 1
        else:
            cur_start = i
            cur_len   = 1

        if cur_len > best_len:       
            best_len   = cur_len
            best_start = cur_start

    result = arr[best_start : best_start + best_len]
    print(f"{best_len}")
    print(f"{result}")

solve()
```

```python
import bisect

def solve():
    arr = list(map(int, input().split()))
    n = len(arr)

    if n == 0:
        print("No")
        return
    tails  = []
    idx    = []
    parent = [-1] * n
    pos    = [0] * n

    for i, x in enumerate(arr):
        p = bisect.bisect_left(tails, x)  
        if p == len(tails):
            tails.append(x)
            idx.append(i)
        else:
            tails[p] = x
            idx[p]   = i

        pos[i] = p
        parent[i] = idx[p - 1] if p > 0 else -1

    best_idx = idx[-1]
    result   = []
    i = best_idx
    while i != -1:
        result.append(arr[i])
        i = parent[i]
    result.reverse()

    print(f"{len(tails)}")
    print(f"{result}")

solve()
```
